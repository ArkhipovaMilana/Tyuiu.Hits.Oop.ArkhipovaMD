<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Student в C#, который будет представлять информацию о
студентах. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.


<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
public class Student 
{
    public string name{get;set;}
    public int age{get;set;}
    public int course{get;set;}
    
    public Student(string name,int age,int course)
    {
        this.name=name;
        this.age=age;
        this.course=course;
    }

    public virtual void GetInfo()
    {
        Console.WriteLine($"Student:{name} ; Age:{age} ; Course:{course}");
    }
    public virtual void Study()
    {
        Console.WriteLine($"{name} is studying...");
    }
    public virtual void TakeExam()
    {
        Console.WriteLine($"{name} is taking an exam...");
    }
}

public class BachelorStudent:Student
{
    public BachelorStudent(string name,int age,int course,string speciality): base (name,age,course)
    {
        this.speciality=speciality;
    }
    public string speciality{get;set;}
    public override void Study()
    {
        Console.WriteLine($"{name} is studying {speciality}...");
    }
}

public class MasterStudent:Student
{
    public MasterStudent(string name,int age,int course,string ScientificAdvisor): base (name,age,course)
    {
        this.ScientificAdvisor=ScientificAdvisor;
    }
    public string ScientificAdvisor;
    public override void TakeExam()
    {
        Console.WriteLine($"{name} is taking an exam with {ScientificAdvisor}...");
    }
}

public class PhDStudent:Student
{
    public PhDStudent(string name,int age,int course,string DissertationTopic): base (name,age,course)
    {
            this.DissertationTopic=DissertationTopic;
    }
    public string DissertationTopic {get;set;}
    public override void Study()
    {
        Console.WriteLine($"{name} is studying {DissertationTopic}...");
    }
}
```

```
BachelorStudent Bob = new BachelorStudent("Bob",18,1,"Computer Science");
MasterStudent Ben = new MasterStudent("Ben",23,5,"Dr.Einstein");
PhDStudent Sven = new PhDStudent("Sven",24,7,"Hash Tables");

Bob.GetInfo();
Bob.Study();
Bob.TakeExam();

Ben.GetInfo();
Ben.Study();
Ben.TakeExam();

Sven.GetInfo();
Sven.Study();
Sven.TakeExam();
```
