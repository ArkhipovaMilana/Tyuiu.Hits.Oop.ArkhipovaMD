```
public abstract class Student 
{
    public string name{get;set;}
    public int age{get;set;}
    public int course{get;set;}
    
    public Student(string name,int age,int course)
    {
        this.name=name;
        this.age=age;
        this.course=course;
    }

    public void GetInfo()
    {
        Console.WriteLine($"Student:{name} ; Age:{age} ; Course:{course}");
    }
    public virtual void Study()
    {
        Console.WriteLine($"{name} is studying...");
    }
    public virtual void TakeExam()
    {
        Console.WriteLine($"{name} is taking an exam...");
    }
}

public interface IStudent
{
    void HalyavaPridi();
}

public class BachelorStudent:Student,IStudent
{
    public string speciality{get;set;}
        public BachelorStudent(string name,int age,int course,string speciality): base (name,age,course)
    {
        this.speciality=speciality;
    }
    public void HalyavaPridi()
    {
        Console.WriteLine("Halyava Pridi!");
    }
}

public class MasterStudent:Student,IStudent
{
    public MasterStudent(string name,int age,int course,string ScientificAdvisor): base (name,age,course)
    {
        this.ScientificAdvisor=ScientificAdvisor;
    }
    public string ScientificAdvisor{get;set;}
    public override void TakeExam()
    {
        Console.WriteLine($"{name} is taking an exam with {ScientificAdvisor}...");
    }
    public void HalyavaPridi()
    {
        Console.WriteLine("Halyava Pridi!!");
    }
}

public class PhDStudent:Student,IStudent
{
    public PhDStudent(string name,int age,int course,string DissertationTopic): base (name,age,course)
    {
            this.DissertationTopic=DissertationTopic;
    }
    public string DissertationTopic {get;set;}
    public override void Study()
    {
        Console.WriteLine($"{name} is studying {DissertationTopic}...");
    }
    public void HalyavaPridi()
    {
        Console.WriteLine("Halyava Pridi!!!");
    }
}
public class StudentCollection<T> where T : Student
{
    private List<T> _students = new List<T>();

    public void Add(T student)
    {
        _students.Add(student);
    }

    public void Remove(T student)
    {
        _students.Remove(student);
    }

    public void DisplayStudents()
    {
        foreach (var student in _students)
        {
            student.GetInfo();
        }
    }
}
StudentCollection<Student> students = new StudentCollection<Student>();
students.Add(new BachelorStudent("gosha",18,1,"Economist"));
students.Add(new MasterStudent("sonya",22,1,"Jason Stethem"));
students.Add(new PhDStudent("zhenya",10,6,"Computers"));
students.DisplayStudents();
```
